```python
sample = sales [
    (sales['item_id'] == "HOBBIES_1_001") &
    (sales['store_id'] == 'CA_1')
].copy()

# Convert wide daily columns into long format
sample_long = sample.melt(
    id_vars=["id", "item_id", "dept_id", "cat_id", "store_id", "state_id"],
    value_vars = day_cols,
    var_name = "d",
    value_name = "sales"
)
# sample_long

sample_long = sample_long.merge(
    calendar[
    [
        "d", "date", "wm_yr_wk", "weekday",
         "event_name_1", "event_type_1",
         "snap_CA", "snap_TX", "snap_WI"]
    ],
    on='d',
    how='left'
)
# Add weekly price
sample_long = sample_long.merge(
    prices,
    on=['store_id', 'item_id', 'wm_yr_wk'],
    how='left'
)
display(sample_long.head(20))
```

```python
# First positive sale
display(sample_long[sample_long['sales']>0]
       [["date", "d", "sales", "sell_price"]]
    .head()
       )
# First available price
display(
    sample_long[sample_long["sell_price"].notna()]
    [["date", "d", "sales", "sell_price"]]
    .head()
)
# Last few observations
display(
    sample_long[
        ["date", "d", "sales", "sell_price"]
    ].tail()
)
```

- First avialable weekly price `2013-07-13(d_897)`	
- first positive sale start from `price 2013-07-18(d_902)`
- after available price information, 1 unit sold
- from `d_1` to `d_896`, price are not included

```python

# Convert price week to first calendar day of that week
week_first_day = (
    calendar.groupby("wm_yr_wk", as_index=False)["d"].first()
)

# Convert d_x to numeric day
week_first_day["day_num"] = (
    week_first_day["d"].str.replace("d_","",regex=False).astype(int)
)
# First price week/day for every store-item
first_price = (prices.groupby(["store_id","item_id"],as_index=False)["wm_yr_wk"]
               .min()
               .merge(week_first_day, on="wm_yr_wk",how="left")
               .rename(columns={"day_num": "first_price_day"})
              )
display(first_price.head())
```
