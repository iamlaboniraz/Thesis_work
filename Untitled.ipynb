# ADI AND CV2 calculate

```python
import numpy as np
import pandas as pd

sales = {
    "Product_A": [3, 0, 0, 2, 0, 0, 4],
    "Product_B": [5, 6, 5, 6, 5, 6, 5],
    "Product_C": [0, 0, 12, 0, 0, 1, 0],
    "Product_D": [2, 15, 1, 17, 3, 14, 2]
}

df = pd.DataFrame(sales)
df
```

```python
def calculate_adi(series):
    total_number_of_periods = len(series)
    non_zeros_count = (series>0).sum()
    if non_zeros_count == 0:
        return np.nan
    return total_number_of_periods/non_zeros_count

for product in df.columns:
    adi = calculate_adi(df[product])
    print(product, "ADI: ",round(adi,2))
```

```python
def calculate_cv2(series): 
    non_zeros = series[series>0]
    if len(non_zeros) < 2:
        return np.nan
    mean = non_zeros.mean()
    variance = non_zeros.var(ddof=0)
    return variance/mean

for product in df.columns:
    cv2 = calculate_cv2(df[product])
    print(product, "cv2: ",round(cv2,3))
```

```python
Product = [3, 0, 0, 2, 0, 0, 4]
mean_value = sum(Product)/len(Product)
mean_value
```

```python
def classify_demand(adi, cv2):
    if np.isnan(adi) or np.isnan(cv2):
        return "Insufficient data"

    if adi < 1.32 and cv2 < 0.49:
        return "Smooth"
    elif adi >= 1.32 and cv2 < 0.49:
        return "Intermittent"
    elif adi < 1.32 and cv2 >= 0.49:
        return "Erratic"
    else:
        return "Lumpy"

results = []

for product in df.columns:
    adi = calculate_adi(df[product])
    cv2 = calculate_cv2(df[product])
    results.append({
        "Product": product,
        "ADI":adi,
        "CV2": cv2,
        "Demand_pattern": classify_demand(adi,cv2)
    })
result_df = pd.DataFrame(results)
result_df
```
